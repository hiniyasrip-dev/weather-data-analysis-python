**WEATHER DATA ANALYSIS AND VISUALIZATION USING PYTHON**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
weather = pd.DataFrame({
    "Date": pd.date_range(start="2026-01-01", periods=30),
    "MinTemp": [18,19,20,21,19,18,20,22,23,21,20,19,18,20,22,23,24,22,21,20,19,18,20,21,23,24,25,23,22,21],
    "MaxTemp": [28,29,31,32,30,28,31,33,34,32,30,29,28,31,33,35,36,34,32,30,29,28,31,32,34,36,37,35,33,32],
    "Rainfall": [0,2,0,5,10,0,3,0,0,8,12,0,4,0,0,15,20,5,0,2,0,7,0,0,10,5,0,3,0,8],
    "Humidity3pm": [65,68,62,70,75,60,67,58,55,72,78,64,69,61,57,80,82,74,63,66,60,71,59,56,76,68,54,62,58,73],
    "Pressure3pm": [1012,1010,1011,1008,1007,1013,1011,1014,1015,1009,1006,1012,1010,1013,1015,1005,1003,1008,1012,1011,1014,1009,1015,1016,1007,1010,1014,1012,1015,1008],
    "WindSpeed3pm": [15,18,12,20,25,14,16,10,12,22,28,15,19,13,11,30,32,24,16,18,14,21,12,10,26,17,9,15,11,23],
    "RainToday": ["No","No","No","Yes","Yes","No","No","No","No","Yes","Yes","No","No","No","No","Yes","Yes","Yes","No","No","No","Yes","No","No","Yes","Yes","No","No","No","Yes"]
})

print("Weather dataset loaded successfully!")
```

Remove Missing Values

```python
weather_clean = weather.dropna()

print("Original dataset size:", weather.shape)
print("Cleaned dataset size:", weather_clean.shape)
```

TEMPERATURE ANALYSIS

Average Maximum Temperature

```python
average_max_temp = np.mean(weather_clean["MaxTemp"])

print("Average Maximum Temperature:", average_max_temp)
```

Maximum and Minimum Temperature

```python
print("Highest Temperature:", weather_clean["MaxTemp"].max())
print("Lowest Temperature:", weather_clean["MinTemp"].min())
```

Temperature Visualization

```python
plt.figure(figsize=(12,5))

plt.plot(weather_clean["MaxTemp"].head(50), label="Maximum Temperature")
plt.plot(weather_clean["MinTemp"].head(50), label="Minimum Temperature")

plt.title("Minimum and Maximum Temperature")
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.grid()

plt.show()
```

💧 HUMIDITY ANALYSIS

Average Humidity

```python
average_humidity = weather_clean["Humidity3pm"].mean()

print("Average Humidity at 3 PM:", average_humidity)
```

Humidity Visualization

```python
plt.figure(figsize=(12,5))

plt.bar(
    range(30),
    weather_clean["Humidity3pm"].head(30)
)

plt.title("Humidity Analysis")
plt.xlabel("Days")
plt.ylabel("Humidity (%)")

plt.show()
```

RAINFALL ANALYSIS

Total Rainfall

```python
total_rainfall = weather_clean["Rainfall"].sum()

print("Total Rainfall:", total_rainfall, "mm")
```

Average Rainfall

```python
average_rainfall = weather_clean["Rainfall"].mean()

print("Average Rainfall:", average_rainfall, "mm")
```

Rainfall Visualization

```python
plt.figure(figsize=(12,5))

plt.bar(
    range(30),
    weather_clean["Rainfall"].head(30)
)

plt.title("Rainfall Analysis")
plt.xlabel("Days")
plt.ylabel("Rainfall (mm)")

plt.show()
```

🌬️ WIND SPEED ANALYSIS

Average Wind Speed

```python
average_wind = weather_clean["WindSpeed3pm"].mean()

print("Average Wind Speed at 3 PM:", average_wind)
```

Wind Speed Visualization

```python
plt.figure(figsize=(12,5))

plt.plot(
    weather_clean["WindSpeed3pm"].head(50)
)

plt.title("Wind Speed Variation")
plt.xlabel("Days")
plt.ylabel("Wind Speed (km/h)")
plt.grid()

plt.show()
```

RAIN TODAY ANALYSIS

Rain Today Count

```python
rain_count = weather_clean["RainToday"].value_counts()

print(rain_count)
```

Rain Today Visualization

```python
rain_count.plot(
    kind="bar"
)

plt.title("Rain Today Distribution")
plt.xlabel("Rain Today")
plt.ylabel("Number of Days")

plt.show()
```


📊 CORRELATION ANALYSIS

Correlation Matrix

```python
correlation = weather_clean[
    ["MinTemp", "MaxTemp", "Rainfall",
     "Humidity3pm", "Pressure3pm", "WindSpeed3pm"]
].corr()

correlation
```

Correlation Heatmap

```python
plt.figure(figsize=(8,6))

plt.imshow(correlation, cmap="coolwarm")

plt.colorbar()

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=45
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.title("Weather Parameters Correlation")

plt.show()
```

📈 TEMPERATURE VS HUMIDITY

Scatter Plot

```python
plt.figure(figsize=(8,5))

plt.scatter(
    weather_clean["MaxTemp"].head(100),
    weather_clean["Humidity3pm"].head(100)
)

plt.title("Temperature vs Humidity")
plt.xlabel("Maximum Temperature (°C)")
plt.ylabel("Humidity (%)")

plt.show()
```

RAINFALL BY TEMPERATURE

```python
plt.figure(figsize=(8,5))

plt.scatter(
    weather_clean["MaxTemp"].head(100),
    weather_clean["Rainfall"].head(100)
)

plt.title("Temperature vs Rainfall")
plt.xlabel("Maximum Temperature (°C)")
plt.ylabel("Rainfall (mm)")

plt.show()
```
